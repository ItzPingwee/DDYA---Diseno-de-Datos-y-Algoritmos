### 5) Análisis de complejidad **[15 pts]**

#### A. 
##### A1. Módulo Max-Heap (Cola de Prioridad):
Justificación: MAX-HEAPIFY(A, i): En su peor caso, un nodo en la raíz debe descender hasta una hoja recorriendo la altura h = $log_2 (n)$ del árbol binario completo, en cada nivel realiza comparaciones constantes (O(1)).

##### A2. BUILD-MAX-HEAP(A): 
Aunque llama a MAX-HEAPIFY $n/2$ veces, la mayoría de los nodos se encuentran cerca de las hojas (La altura es pequeña), sumando el trabajo por niveles: $\sum_{h=0}^{\log n} \frac{n}{2^{h+1}}$.

O(h) = O(n)

##### A3. EXTRACT-MAX e INSERT: 
Ambas modifican un extremo del árbol (raíz o última hoja) y restauran la propiedad recorriendo a lo sumo la altura del árbol, lo que toma O(log n).

#### Función T(n): 
- Para MAX-HEAPIFY / EXTRACT-MAX / INSERT:
$$T(n) = c_1 \log_2 n + c_2 \implies O(\log n)$$

- Para BUILD-MAX-HEAP:
$$T(n) = \sum_{i=1}^{n/2} c \cdot h_i \le 2cn \implies O(n)$$

#### Mejor y peor caso: 

##### Mejor caso (O(1) en inserción / O(n) en build): 
Al insertar un elemento menor que su padre, no requiere intercambios (1 comparación). En BUILD-MAX-HEAP, si el arreglo ya cumple la propiedad de Max-Heap, solo verifica los hijos sin realizar intercambios (O(n)).




##### Peor caso (O(Log n) en inserción/extracción / O(n) en build): 
Al insertar una prioridad mayor a todas las existentes (como T07 con prioridad 12), el elemento sube desde la última hoja hasta la raíz recorriendo $log_2 n$ niveles.




#### B. Módulo Algoritmo de Kruskal (MST)
Justificación: Dado un grafo con V vértices y E = $\frac{V(V-1)}{2}$ aristas:
Generar todas las parejas toma $O(V^2) = O(E)$.

Ordenar las E aristas por peso toma $(E \log E)$.

Recorrer las aristas aplicando Find y Union toma $O(E \cdot \alpha(V))$, donde $\alpha(V)$ es la inversa de la función de Ackermann (Donde la funcion de ackerman es aquella que crece muy rapido, en este caso su crecimiento es muy lento) ($\alpha(V) \le 4$ en la práctica).

#### Función $T(E, V)$:

$$T(E, V) = c_1 E + c_2 E \log_2 E + c_3 E \alpha(V) \implies O(E \log E)$$

(Como $E \le V^2$, $\log E \le 2 \log V$, por lo que también se expresa como $O(E \log V)$).

#### Mejor y peor caso:
##### Mejor caso ($O(E \log E)$):
El ordenamiento inicial de las aristas domina el tiempo asintótico incluso si las primeras $V-1$ aristas ordenadas no forman ciclos y detienen el ciclo principal tempranamente.




##### Peor caso ($O(E \log E)$): 
Además de ordenar todas las aristas, se deben evaluar casi todas las $E$ aristas antes de completar las $V-1$ conexiones válidas del MST.

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# PUNTO 1 Y 2: COLA DE PRIORIDAD MÁXIMA (MAX-HEAP)

def padre_idx(i):
    return (i - 1) // 2

def izq_idx(i):
    return 2 * i + 1

def der_idx(i):
    return 2 * i + 2

def max_heapify(heap, i):
    """Mantiene la propiedad de Max-Heap hacia abajo desde el índice i."""
    n = len(heap)
    l = izq_idx(i)
    r = der_idx(i)
    largest = i

    if l < n and heap[l]["prioridad"] > heap[largest]["prioridad"]:
        largest = l
    if r < n and heap[r]["prioridad"] > heap[largest]["prioridad"]:
        largest = r

    if largest != i:
        heap[i], heap[largest] = heap[largest], heap[i]
        max_heapify(heap, largest)

def build_max_heap(tareas):
    """Construye y retorna un Max-Heap a partir de un listado de tareas."""
    heap = [t.copy() for t in tareas]
    n = len(heap)
    for i in range((n // 2) - 1, -1, -1):
        max_heapify(heap, i)
    return heap

def extract_max(heap):
    """Extrae y retorna (tarea_maxima, nuevo_heap)."""
    if len(heap) == 0:
        raise IndexError("El Heap está vacío.")

    nuevo_heap = [t.copy() for t in heap]
    max_item = nuevo_heap[0]
    last_item = nuevo_heap.pop()

    if len(nuevo_heap) > 0:
        nuevo_heap[0] = last_item
        max_heapify(nuevo_heap, 0)
    return max_item, nuevo_heap

def insert_tarea(heap, tarea):
    """Inserta una nueva tarea y retorna el Heap actualizado."""
    nuevo_heap = [t.copy() for t in heap]
    nuevo_heap.append(tarea.copy())
    i = len(nuevo_heap) - 1

    while i > 0 and nuevo_heap[padre_idx(i)]["prioridad"] < nuevo_heap[i]["prioridad"]:
        p = padre_idx(i)
        nuevo_heap[i], nuevo_heap[p] = nuevo_heap[p], nuevo_heap[i]
        i = p
    return nuevo_heap

def mostrar_arreglo(heap, titulo=""):
    if titulo:
        print(f"\n--- {titulo} ---")
    for idx, t in enumerate(heap):
        print(f"  [{idx}] {t['codigo']} (Prioridad:{t['prioridad']} - {t['nombre']})")

def graficar_arbol(heap, titulo="Árbol Binario del Max-Heap"):
    """Dibuja la estructura jerárquica del Max-Heap usando Matplotlib."""
    n = len(heap)
    if n == 0:
        return

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.set_title(titulo, fontsize=13, fontweight="bold", pad=15)
    ax.axis("off")

    posiciones = {}
    niveles = math.floor(math.log2(n)) + 1

    for i in range(n):
        nivel = math.floor(math.log2(i + 1))
        pos_en_nivel = i - (2**nivel - 1)
        total_en_nivel = 2**nivel
        x = (pos_en_nivel + 0.5) / total_en_nivel
        y = 1.0 - (nivel + 0.5) / niveles
        posiciones[i] = (x, y)

    for i in range(n):
        x1, y1 = posiciones[i]
        for hijo in (izq_idx(i), der_idx(i)):
            if hijo < n:
                x2, y2 = posiciones[hijo]
                ax.plot([x1, x2], [y1, y2], color="#555555", linewidth=1.8, zorder=1)

    for i, tarea in enumerate(heap):
        x, y = posiciones[i]
        ax.scatter(x, y, s=2200, color="#1f77b4", edgecolors="black", linewidth=1.5, zorder=2)
        etiqueta = f"{tarea['codigo']}\nP: {tarea['prioridad']}\n{tarea['nombre'][:12]}"
        ax.text(x, y, etiqueta, ha="center", va="center", color="white",
                fontsize=8, fontweight="bold", zorder=3)

    plt.tight_layout()
    plt.show()


# Ejecución Punto 1: Datos iniciales y BUILD-MAX-HEAP
tareas_iniciales = [
    {"codigo": "T01", "nombre": "Configurar servidor", "prioridad": 6},
    {"codigo": "T02", "nombre": "Reparar sensor",      "prioridad": 10},
    {"codigo": "T03", "nombre": "Actualizar software", "prioridad": 4},
    {"codigo": "T04", "nombre": "Revisar cámara",      "prioridad": 8},
    {"codigo": "T05", "nombre": "Generar respaldo",    "prioridad": 3},
    {"codigo": "T06", "nombre": "Configurar alarma",   "prioridad": 7},
]

print("PUNTO 1: CONSTRUCCIÓN DEL MAX-HEAP")
mostrar_arreglo(tareas_iniciales, "Arreglo inicial")
heap_actual = build_max_heap(tareas_iniciales)
mostrar_arreglo(heap_actual, "Max-Heap Resultante (BUILD-MAX-HEAP)")
graficar_arbol(heap_actual, "Punto 1: Árbol correspondiente al Max-Heap inicial")

# Ejecución Punto 2: Procesar 2 tareas e insertar T07
print("PUNTO 2: PROCESAR TAREAS (EXTRACT-MAX) E INSERTAR T07")

for rep in range(1, 3):
    procesada, heap_actual = extract_max(heap_actual)
    print(f"\nExtracción {rep}. - Tarea procesada: {procesada['codigo']} | {procesada['nombre']} (Prioridad: {procesada['prioridad']})")
    mostrar_arreglo(heap_actual, f"Heap resultante tras la extracción #{rep}")

nueva_tarea = {"codigo": "T07", "nombre": "Falla en sistema contra incendios", "prioridad": 12}
heap_actual = insert_tarea(heap_actual, nueva_tarea)
print(f"\nInserción de tarea de emergencia: {nueva_tarea['codigo']} | {nueva_tarea['nombre']} (Prioridad: {nueva_tarea['prioridad']})")
mostrar_arreglo(heap_actual, "Heap resultante tras insertar T07")
graficar_arbol(heap_actual, "Punto 2: Max-Heap tras dos extracciones e inserción de T07")


# PUNTO 3, 4 Y 5: DISPOSITIVOS, KRUSKAL MANUAL Y VISUALIZACIÓN DEL MST
dispositivos = {
    "A": {"nombre": "Servidor", "x": 2,  "y": 2},
    "B": {"nombre": "Cámara",   "x": 7,  "y": 3},
    "C": {"nombre": "Sensor",   "x": 4,  "y": 7},
    "D": {"nombre": "Alarma",   "x": 10, "y": 8},
    "E": {"nombre": "Router",   "x": 13, "y": 3},
    "F": {"nombre": "Panel",    "x": 15, "y": 7},
}

print("================== Parte 2 ==================")
print("Posiciones inciales:")

# Punto 3: Gráfica inicial de posiciones
fig, ax = plt.subplots(figsize=(8, 5))
for id_disp, info in dispositivos.items():
    ax.scatter(info["x"], info["y"], s=250, color="#ff7f0e", edgecolors="black", zorder=3)
    ax.text(info["x"], info["y"] + 0.35, f"{id_disp} ({info['nombre']})\n({info['x']},{info['y']})",
            ha="center", va="bottom", fontsize=9, fontweight="bold")

ax.set_title("Punto 3: Ubicación espacial de los dispositivos", fontsize=12, fontweight="bold")
ax.set_xlabel("Coordenada X")
ax.set_ylabel("Coordenada Y")
ax.set_xlim(0, 17)
ax.set_ylim(0, 10.5)
ax.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

def distancia_euclidiana(p1, p2):
    return math.sqrt((p2["x"] - p1["x"])**2 + (p2["y"] - p1["y"])**2)

aristas = []
ids = list(dispositivos.keys())
for i in range(len(ids)):
    for j in range(i + 1, len(ids)):
        u, v = ids[i], ids[j]
        peso = distancia_euclidiana(dispositivos[u], dispositivos[v])
        aristas.append((peso, u, v))

# Punto 4: Implementación manual de Kruskal
def crear_conjuntos(vertices):
    padre = {v: v for v in vertices}
    rango = {v: 0 for v in vertices}
    return padre, rango

def encontrar_raiz(padre, item):
    if padre[item] != item:
        padre[item] = encontrar_raiz(padre, padre[item])
    return padre[item]

def unir_conjuntos(padre, rango, u, v):
    raiz_u = encontrar_raiz(padre, u)
    raiz_v = encontrar_raiz(padre, v)
    if raiz_u != raiz_v:
        if rango[raiz_u] < rango[raiz_v]:
            padre[raiz_u] = raiz_v
        elif rango[raiz_u] > rango[raiz_v]:
            padre[raiz_v] = raiz_u
        else:
            padre[raiz_v] = raiz_u
            rango[raiz_u] += 1
        return True
    return False

def kruskal_manual(vertices, lista_aristas):
    aristas_ordenadas = sorted(lista_aristas, key=lambda x: x[0])
    padre, rango = crear_conjuntos(vertices)
    mst = []
    costo_total = 0.0

    for peso, u, v in aristas_ordenadas:
        if unir_conjuntos(padre, rango, u, v):
            mst.append((u, v, peso))
            costo_total += peso
            if len(mst) == len(vertices) - 1: # Max n-1 aristas
                break

    return mst, costo_total

mst_seleccionado, longitud_total = kruskal_manual(ids, aristas)

print("PUNTO 4: ÁRBOL DE EXPANSIÓN MÍNIMA (ALGORITMO DE KRUSKAL)")
print("Conexiones seleccionadas:")
for u, v, peso in mst_seleccionado:
    print(f"  {u} - {v}: {peso:.2f}  ({dispositivos[u]['nombre']} <-> {dispositivos[v]['nombre']})")
print(f"\nLongitud total de cable: {longitud_total:.2f}")

# Punto 5: Visualización final del MST
fig, ax = plt.subplots(figsize=(9, 5.5))

for u, v, peso in mst_seleccionado:
    x1, y1 = dispositivos[u]["x"], dispositivos[u]["y"]
    x2, y2 = dispositivos[v]["x"], dispositivos[v]["y"]
    ax.plot([x1, x2], [y1, y2], color="#2ca02c", linewidth=2.5, zorder=2)

    mx, my = (x1 + x2) / 2, (y1 + y2) / 2
    ax.text(mx, my, f"{peso:.2f}", fontsize=9, fontweight="bold", color="#1b5e20",
            bbox=dict(boxstyle="round,pad=0.2", facecolor="white", edgecolor="#2ca02c", alpha=0.9),
            zorder=4)

for id_disp, info in dispositivos.items():
    ax.scatter(info["x"], info["y"], s=300, color="#1f77b4", edgecolors="black", linewidth=1.5, zorder=3)
    ax.text(info["x"], info["y"] + 0.35, f"{id_disp} ({info['nombre']})",
            ha="center", va="bottom", fontsize=9.5, fontweight="bold", zorder=5)

ax.set_title(f"Punto 5: Árbol de Expansión Mínima (MST) - Longitud total de cable: {longitud_total:.2f}",
             fontsize=12, fontweight="bold")
ax.set_xlabel("Coordenada X")
ax.set_ylabel("Coordenada Y")
ax.set_xlim(0, 17)
ax.set_ylim(0, 10.5)
ax.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
